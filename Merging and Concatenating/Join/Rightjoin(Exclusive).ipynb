```python
import pandas as pd
```

```python
df1=pd.DataFrame({'id':['A','B','C','D'],
                   'age':[30,23,25,22]})
df2=pd.DataFrame({'id':['C','D','E','F'],
                   'job':['Doctor','Statistician',
                        'Accountant','Developer']})
```

```python
df1.merge(df2,on='id',how='outer', indicator=True)
```

```python
df1.merge(df2,on='id',how='right', 
          indicator=True).query(" _merge=='right_only'")
```

```python
df_movies=pd.read_csv('IMDb movies.csv',low_memory=False)
df_ratings=pd.read_csv('IMDb ratings.csv')
```

```python
# make a copy of the df_ratings dataframe 
df_rating_copy=df_ratings.copy()
```

```python
# set the first 100 values if 'imdb_title_id' column as 'tt1234567890'
for index in df_rating_copy.index:
     if index<1000:
         df_rating_copy.loc[index,'imdb_title_id']='tt1234567890'
     
```

```python
df_rating_copy
```

```python
# merge df_movies and df_rating_copy
right_only=df_movies.merge( df_rating_copy,on='imdb_title_id',how='outer',
               indicator=True).query("_merge=='right_only'")
```
