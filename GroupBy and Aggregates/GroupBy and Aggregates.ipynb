```python
import pandas as pd 
```

```python
df=pd.read_csv('Car_sales.csv')
df
```

```python
# selecting columns
df=df[['Manufacturer','Sales_in_thousands',
                 'Vehicle_type','Price_in_thousands',
                 'Engine_size','Horsepower','Fuel_capacity']]
```

```python
df
```

```python
# calculate the number of unique values
df.nunique()
```

# agg()

```python
# calculate the sum of columns
df.agg('sum')
```

```python
#calculate the mean of numeric columnd
df.agg('mean',numeric_only=True)
```

```python
# Different aggregations per column:calculate the sum and mean of 'Sales_in_thousands'
#and sum and max of 'Price_in_thousands'
df.agg({'Sales_in_thousands':['sum','mean'],
         'Price_in_thousands':['sum','max']})
```

```python
#aggregate over the columns
df[['Sales_in_thousands','Price_in_thousands']].agg('sum',axis=1)
```

```python
# equivalent (NaN behaves differently)
df['Sales_in_thousands']+df['Price_in_thousands']
```

```python
# aggregate different functions and reaname the index of resulting dataframe
df.agg(x=('Sales_in_thousands','sum'),
       y=('Price_in_thousands','sum'))
```

# The Split-Apply-Combine Strategy

```python
# Split data into separate groups
passenger_filter=df['Vehicle_type']=='Passenger'
car_filter=df['Vehicle_type']=='Car'
#df['Vehicle_type'].value_counts()
```

```python
# apply an operation
passenger_avg=df[passenger_filter] ['Sales_in_thousands'].mean()
car_avg=df[car_filter]['Sales_in_thousands'].mean()
```

```python
# combine results
pd.DataFrame({'Vehicle_type':['Car','Passenger'],
              'Sales_in_thousands':[car_avg,passenger_avg]})
```
