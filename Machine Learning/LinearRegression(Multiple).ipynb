```python
import pandas as pd
```

```python
df_boston=pd.read_csv('Boston House Prices.csv')
df_boston
```

# Mulitple linear regression

```python
import statsmodels.api as sm
```

# Define the dependent and indepnedent variable

```python
# dependent varaiable
y=df_boston['Value']
# independent variable
X=df_boston[['Rooms','Distance']]
```

# Making the regression: Adding a constant and fitting the model

```python
# adding a constant
sm.add_constant(X)
# fitting the model
lm=sm.OLS(y,X).fit()
```

```python
lm.summary()
```
