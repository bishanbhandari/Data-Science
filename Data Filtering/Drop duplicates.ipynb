# drop_duplicates()

```python
import pandas as pd
```

```python
df_laptop=pd.read_csv('laptop_price.csv',encoding='latin1')
```

```python
df_laptop.head(5)
```

```python
#drop duplicates in two or more columns 
df_laptop.drop_duplicates(['Company']) [['Company','Price_euros']].value_counts('Company')
```

```python
#sort dataframe ascending by company and price
#cheapst first and most expensive last
df_laptop=df_laptop.sort_values(['Company','Price_euros'])
```

```python
#cheapest:keep='first'
df_laptop.drop_duplicates(['Company'],keep='first')[['Company','Price_euros']]
```

```python
#expensive keep ='last'
df_laptop.drop_duplicates(['Company'],keep='last')[['Company','Price_euros']]
```
