# Creating a conditional column from More than 2 choices:np.select()

```python
import numpy as np
```

```python
#create an array based on mulitple price tiers (+2 choices)
#create "conditions" and "values" variable
conditions =[
    df_laptop['Price_euros']>3000,
    (df_laptop['Price_euros']>2000) & (df_laptop['Price_euros']<=3000),
    (df_laptop['Price_euros']>800)&(df_laptop['Price_euros']<=2000),
    df_laptop['Price_euros']<=800
]
values=['Too Expensive','Expensive','Affordable','Cheap']
```

```python
#set it to a new column
df_laptop['Price_tiers']=np.select(conditions ,values)
```

```python
df_laptop
```

```python
#count values inside "Price_tiers" column
df_laptop.value_counts('Price_tiers')
```

```python
#create an array based on multiple screen size tiers (+2 choices)
#Too Big>16, Big>=14, Small>12 Too Small<13.5
#Create "conditions" and "values" variables 
conditions=[
    df_laptop['Inches']>16,
    (df_laptop['Inches']>=14)&(df_laptop['Inches']<=16),
    (df_laptop['Inches']>12)&(df_laptop['Inches']<=14),
    df_laptop['Inches']<=12
]
values=['Too Big','Big','Small','Too Small']
```

```python
#Set it to a new column
df_laptop['Screen_size']=np.select(conditions,values)
```
