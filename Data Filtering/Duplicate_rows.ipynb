# Duplicated()

```python
# find duplicates in 1 column/series
df_laptop.duplicated('laptop_ID')
```

```python
# show elements in dataframe with duplicates in column 'laptop_ID'
df_laptop[df_laptop.duplicated('laptop_ID')]
```

```python
# Duplicates in two or more column'
df_laptop.duplicated(['Product'	,'TypeName','Inches'])
```

```python
#show all the values duplicted (except for the first- keep='first')
df_laptop[df_laptop.duplicated(['Product'	,'TypeName','Inches'])]
```

# Example

```python
#sort dataframe ascending by company and price(cheapest first and most expensive last)
df_laptop=df_laptop.sort_values(['Company','Price_euros'])
df_laptop
```

```python
#check all the value categories in "Company column"
df_laptop.value_counts(['Company'])
```

```python
# duplicated vlues in "Company" columns (default-> keep='first'
duplicated_first=df_laptop.duplicated('Company',keep='first')
```

```python
# show dataframe with the duplicated values in "company" column
df_laptop[duplicated_first]
#keep='first'(cheapest laptop per company)
#show dataframe with non duplicated values in "Company" column
df_laptop[~duplicated_first] [['Company', 'Price_euros']]
#check all the value categories
df_laptop[~duplicated_first].value_counts('Company')
```

```python
# keep = 'last'
duplicated_last=df_laptop.duplicated('Company',keep='last')
# keep= false
duplicated_false=df_laptop.duplicated('Company',keep=False)
```

```python
#keep='last'(most expensive laptop per company)
#show dataframe with non duplicated values in "Company" column
df_laptop[~duplicated_last][['Company','Price_euros']]
```

```python
#keep=false
#show dataframe with non duplicated values in "Company" column
df_laptop[~duplicated_false]
```
