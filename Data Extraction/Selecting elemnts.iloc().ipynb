```python
import pandas as pd
```

```python
df=pd.read_csv('players_20.csv')
df.set_index('short_name',inplace=True)
df=df[['long_name',	'age', 'dob',	'height_cm',	'weight_kg',	'nationality', 'club']]
```

```python
df
```

# Selecting with a single value
  iloc[row_position,column_position]

```python
# get the height of Cristiano Ronaldo
df.iloc[1,3]
```

```python
# get the weight of Cristiano Ronaldo
df.iloc[1,4]
```

```python
#get all rows inside the 'height_cm column'
df.iloc[:,3]
```

```python
#get all column that correspond to the index 'Cristiano Ronaldo'
df.iloc[1,:]
```

# Selecting with a list of values 

```python
#get all the data about Cristiano Ronaldo and L. Messi
df.iloc[[0,1]]
```

```python
# get the height of Cristiano Ronaldo and L. Messi
df.iloc[[0,1],3]
```

```python
# get height and weight of c ronaldo
df.iloc[1,[3,4]]
```

```python
# get height and weight of c ronaldo and l messi
df.iloc[[0,1],[3,4]]
```

# Selecting a range of data with a slice
  start:stop(note that the start is included but the stop is excluded)


```python
#slice column lables age/club
players=[0,1]
df.iloc[players,1:7]
```

```python
#slice index lables(top1 to top10 players
df.iloc[0:10,1:5]
```

# Selecting with conditions
  iloc cannot accept the boolean series, but only a boolean list. we have to use the list() function to convert a series into a boolean list

```python
#one condition:select player with height above 180cm
df.iloc[list(df['height_cm']>180),1:5]
```

```python
#multiple conditions : select a player with height above 180cm from portugal
df.iloc[list((df['height_cm'] > 180) & (df['nationality']=="Portugal")),:]
```
