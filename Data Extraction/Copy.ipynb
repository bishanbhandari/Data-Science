```python
import pandas as pd 
```

```python
df=pd.read_csv('players_20.csv')
#indexing 
df.set_index('short_name',inplace=True)
#select column
df=df[['long_name',	'age', 'dob',	'height_cm',	'weight_kg',	'nationality', 'club']]
```

# Copy 
  deep=True
  

```python
# deep=True by default(modification to the data or indices of the copy) df_copy(independent copy)
df_copy=df.copy()
```

```python
#update value in original dataframe 
df.loc['L. Messi','height_cm']=180
```

```python
df
df_copy
```

# deep=False

```python
# deep= false (any changes to the data of the original will be reflected
# in the shallow copy and vice versa)
df_shallow_copy=df.copy(deep=False)
```

```python
#update value in original dataframe
df.loc['Cristiano Ronaldo','height_cm']=200
```

```python
#shallow copy vs origonam dataframe
df_shallow_copy
df
```

# Simple assignment

```python
# make a copy 
df_new_copy=df
```

```python
# update value in original dataframe
df.loc['Neymar Jr','height_cm']=180
```

```python
# copy vs original dataframe
df
df_new_copy
```
