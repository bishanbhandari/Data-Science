```python
import pandas as pd
```

```python
df=pd.read_csv('players_20.csv')
df.set_index('short_name',inplace=True)
df=df[['long_name',	'age', 'dob',	'height_cm',	'weight_kg',	'nationality', 'club']]
```

```python
df
```

# selecting with a single value
 loc[row_label,column_label]

```python
# get all data about C.Ronaldo
df.loc['Cristiano Ronaldo']
```

```python
df.loc['Cristiano Ronaldo','height_cm']
```

```python
# get all rows inside the 'height_cm' column
df.loc[:,'height_cm']
```

```python
#get all columns that correspond to the index 'C.Ronaldo'
df.loc['Cristiano Ronaldo',:]
```

# 2 Selecting with the list of values

```python
#get all data about c.ronaldo and l.messi
df.loc[['Cristiano Ronaldo','L. Messi']]
```

```python
#get height of both players
df.loc[['L. Messi','Cristiano Ronaldo'],'height_cm']
```

```python
#get the  height and weight of c ronaldo
df.loc['Cristiano Ronaldo',['height_cm','weight_kg']]
```

```python
#get height and weight of both players
df.loc[['L. Messi','Cristiano Ronaldo'],['height_cm','weight_kg']]
```

# Selecting a range of data with a slice
  start:stop:step(note that contrary to usual python slices,both the start and stop are included)

```python
#slice column lables
players=['L. Messi','Cristiano Ronaldo']
df.loc[players,'age':'club']
```

```python
#slice index lables
columns=['age','dob','height_cm','weight_kg']
#get top1 and top10 player name
df.index[:10]
df.loc['L. Messi':'M. Salah',columns]
```

```python
# 4 Selecting with conditions:select player with height above 180cm
columns=['age','dob','height_cm','weight_kg']
df.loc[df['height_cm']>180,columns]
```

```python
#multiple conditions: select player with height above 180cm from Portugal
df.loc[(df['height_cm']>180)&(df['nationality']=="Portugal"),:]
```
