# The value_counts() Method

# 1 Value counts

```python
#counting gender elements 
#len fucntion 
len(df_exam['gender'])
# .count()method
df_exam['gender'].count()
```

```python
#counting gender elements by category
df_exam['gender'].value_counts()
```

```python
#return the relative frequency(divide all the values by the sum of values)
df_exam['gender'].value_counts(normalize=True)
```

```python
#counting "parental level of education" elements by catogery
df_exam['parental level of education'].value_counts()
```

```python
df_exam['parental level of education'].value_counts(normalize=True)
```

# Sort a DataFrame

```python
#sort by one column
df_exam.sort_values('math score')
```

```python
#sort descending by one column
df_exam.sort_values('math score', ascending=False)
```

```python
#sort descednding by multiple column
df_exam.sort_values(['math score','reading score'],ascending=False)
```

```python
#sort descednding by multiple column and update dataframe
df_exam.sort_values(['math score','reading score'],ascending=False, inplace=True)
df_exam
```

```python
#sort descending with a key function
df_exam.sort_values('race/ethnicity',ascending=True, key=lambda col:col.str.lower())
```

# Create Index

```python
import random
```

```python
#creating non repetitive values for the index
new_index=np.arange(0,1000)
```

```python
#shuffling the indexes
random.shuffle(new_index)
```

```python
#create a new column with new index 
df_exam['new_index']=new_index
```

# Set Index

```python
# set new index column as index
df_exam.set_index('new_index',inplace=True)
```

```python
df_exam
```

```python
# sort dataframe by new index
df_exam.sort_index()
```

```python
df_exam.sort_index(ascending=False)
```

# Rename columns and index with rename()

```python
# rename column and overwrite values of the dataframe 
df_exam=df_exam.rename(columns={'gender':'Gender'})
df_exam.head()
```

```python
#rename multiple column and update the dataframe with inplace argument 
df_exam.rename(
    columns={
        'math score': 'math_score',
        'reading score': 'reading_score'
    },
    inplace=True
)
```

```python
df_exam.head(2)
```

# Rename Index

```python
#rename index 0,1 and 2 and update the dataframe
df_exam.rename(index={0:'A',1:'B',2:'C'},inplace=True)
```
